```python
import os, sys, json, numpy as np
import cv2
```

```python
image_path = "/datassd1/users/chandrakanth/portfolio/chand/chandrakanth-gudavalli-protfolio/notebooks/unnamed.png"
chandu_image = cv2.imread(image_path)
```

```python
icons_dir = "/datassd1/users/chandrakanth/portfolio/chand/chandrakanth-gudavalli-protfolio/public/icons"
images_list = [item for item in os.listdir(icons_dir) if item.endswith(".png") and "ms-icon" in item]
images_list.sort()
print(images_list)

# image_name = images_list[0]
for image_name in images_list:
    icon_path = os.path.join(icons_dir, image_name)
    icon = cv2.imread(icon_path)
    # reshape chandu_image to the size of icon
    chandu_image_resized = cv2.resize(chandu_image, (icon.shape[1], icon.shape[0]))
    # save chandu_image_resized to icon_path
    cv2.imwrite(icon_path, chandu_image_resized)
    print(f"Saved {icon_path}")
```
